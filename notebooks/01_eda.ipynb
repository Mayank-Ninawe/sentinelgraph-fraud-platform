```python
import pandas as pd

df = pd.read_parquet(
    r"D:\sentinelgraph-fraud-platform\data\processed\transactions.parquet"
)

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplicate transaction IDs:")
print(df["transaction_id"].duplicated().sum())

print("\nFraud distribution:")
print(df["is_fraud"].value_counts().sort_index())
```

```python
import pandas as pd

raw = pd.read_csv(
    r"D:\sentinelgraph-fraud-platform\data\raw\paysim.csv"
)

processed = pd.read_parquet(
    r"D:\sentinelgraph-fraud-platform\data\processed\transactions.parquet"
)

print("Raw shape:", raw.shape)
print("Processed shape:", processed.shape)

print("Raw duplicate rows:", raw.duplicated().sum())
print("Raw duplicate transaction positions:", raw.index.duplicated().sum())

print("Raw step range:", raw["step"].min(), raw["step"].max())
print(
    "Processed time range:",
    processed["event_time"].min(),
    processed["event_time"].max(),
)

print("Raw fraud count:")
print(raw["isFraud"].value_counts())

print("Processed fraud count:")
print(processed["is_fraud"].value_counts())
```

```python
raw_counts = raw.groupby(
    ["step", "nameOrig", "nameDest", "amount", "type", "isFraud"],
    dropna=False,
).size()

processed_counts = processed.groupby(
    [
        "event_time",
        "sender_id",
        "receiver_id",
        "amount",
        "transaction_type",
        "is_fraud",
    ],
    dropna=False,
).size()

print("Raw grouped rows:", raw_counts.sum())
print("Processed grouped rows:", processed_counts.sum())
```
